```python
import pandas as pd
import numpy as np
import holidays
import os

# 📁 Load Excel from your local machine
file_path = r"D:\RA Work\Projects\AI Works\DCC_Call History.xlsx"
df = pd.read_excel(file_path)

# 🧹 Preprocess
df['Date_Time'] = pd.to_datetime(df['Date_Time'])
df = df.sort_values(by='Date_Time')

# 🎉 Holiday flags (US + Canada)
years = df['Date_Time'].dt.year.unique()
us_holidays = holidays.US(years=years)
ca_holidays = holidays.CA(years=years)
combined_holidays = set(us_holidays.keys()) | set(ca_holidays.keys())
df['is_holiday'] = df['Date_Time'].dt.date.isin(combined_holidays).astype(int)
df['holiday_name'] = df['Date_Time'].dt.date.map(lambda d: us_holidays.get(d) or ca_holidays.get(d))

# 🧩 Fill missing 30-min intervals per queue
def fill_queue_gaps(df):
    full_index = pd.DataFrame()
    for q in df['Queue'].unique():
        df_q = df[df['Queue'] == q].copy()
        full_range = pd.date_range(start=df_q['Date_Time'].min(), end=df_q['Date_Time'].max(), freq='30min')
        full_df_q = pd.DataFrame({'Date_Time': full_range})
        full_df_q['Queue'] = q
        full_df_q = pd.merge(full_df_q, df_q, on=['Date_Time', 'Queue'], how='left')
        full_index = pd.concat([full_index, full_df_q], ignore_index=True)
    return full_index

df = fill_queue_gaps(df)
df['Calls'] = df['Calls'].fillna(0)

# 🕒 Time features
df['day_of_week'] = df['Date_Time'].dt.dayofweek
df['hour'] = df['Date_Time'].dt.hour
df['minute'] = df['Date_Time'].dt.minute
df['day_of_year'] = df['Date_Time'].dt.dayofyear
df['week_of_year'] = df['Date_Time'].dt.isocalendar().week.astype(int)

# 🔢 One-hot encode queues
df = pd.get_dummies(df, columns=['Queue'], drop_first=False, prefix='Queue')
queue_cols = [col for col in df.columns if col.startswith('Queue_')]
```

Define Evaluation Metrics

```python
import numpy as np
from sklearn.metrics import mean_absolute_error, mean_squared_error

def evaluate(y_true, y_pred, verbose=True):
    y_true = np.array(y_true)
    y_pred = np.array(y_pred)

    mae = mean_absolute_error(y_true, y_pred)
    rmse = mean_squared_error(y_true, y_pred) ** 0.5
    non_zero_mask = y_true != 0

    mape = (
        np.mean(np.abs((y_true[non_zero_mask] - y_pred[non_zero_mask]) / y_true[non_zero_mask])) * 100
        if np.any(non_zero_mask) else np.nan
    )

    if verbose:
        print(f"📊 MAE:  {mae:.2f}")
        print(f"📊 RMSE: {rmse:.2f}")
        print(f"📊 MAPE: {mape:.2f}%")

    return mae, rmse, mape
```

### Prophet Model

```python
from prophet import Prophet
import holidays

def run_prophet(df_queue, forecast_steps=48, verbose=True):
    # Prepare data
    df_prophet = df_queue[['Date_Time', 'Calls']].rename(columns={'Date_Time': 'ds', 'Calls': 'y'}).copy()

    # Generate US + Canada holidays
    years = df_prophet['ds'].dt.year.unique()
    us_holidays = holidays.US(years=years)
    ca_holidays = holidays.CA(years=years)
    combined_holidays = set(us_holidays.keys()) | set(ca_holidays.keys())

    holiday_df = pd.DataFrame({
        'holiday': 'US_CA_Holiday',
        'ds': pd.to_datetime(list(combined_holidays)),
        'lower_window': 0,
        'upper_window': 1
    })

    # Fit Prophet model
    model = Prophet(holidays=holiday_df)
    model.fit(df_prophet)

    # Forecast future steps
    future = model.make_future_dataframe(periods=forecast_steps, freq='30min')
    forecast = model.predict(future)

    if verbose:
        print(f"✅ Prophet model trained on {len(df_prophet)} rows")
        print(f"📈 Forecasting {forecast_steps} steps ahead")

    return forecast[['ds', 'yhat']].set_index('ds'), df_prophet.set_index('ds')
```

### ARIMA Model

```python
from statsmodels.tsa.arima.model import ARIMA

def run_arima(df_queue, forecast_steps=48, verbose=True):
    df_queue = df_queue.copy()

    # Ensure datetime index and frequency
    df_queue['Date_Time'] = pd.to_datetime(df_queue['Date_Time'])
    ts = df_queue.set_index('Date_Time')['Calls']
    ts = ts.asfreq('30min')  # Explicit frequency for ARIMA

    # Fit ARIMA model
    model = ARIMA(ts, order=(1, 1, 1))
    results = model.fit()

    # Forecast future steps
    forecast = results.forecast(steps=forecast_steps)

    if verbose:
        print(f"✅ ARIMA model trained on {len(ts)} time steps")
        print(f"📈 Forecasting {forecast_steps} steps ahead")

    return forecast, ts[-forecast_steps:]
```

### XGBoost & Random Forest

```python
from xgboost import XGBRegressor
from sklearn.ensemble import RandomForestRegressor

def run_ml_model(df_queue, model_type='xgb', verbose=True):
    df_queue = df_queue.copy()

    # Lag features
    df_queue['lag_1'] = df_queue['Calls'].shift(1)
    df_queue['lag_24'] = df_queue['Calls'].shift(48)
    df_queue.dropna(inplace=True)

    # Feature selection
    features = ['hour', 'day_of_week', 'lag_1', 'lag_24']
    if 'is_holiday' in df_queue.columns:
        features.append('is_holiday')

    X = df_queue[features]
    y = df_queue['Calls']

    # Train-test split
    split = int(len(X) * 0.8)
    X_train, X_test = X.iloc[:split], X.iloc[split:]
    y_train, y_test = y.iloc[:split], y.iloc[split:]

    # Model selection
    model = XGBRegressor() if model_type == 'xgb' else RandomForestRegressor()
    model.fit(X_train, y_train)
    y_pred = model.predict(X_test)

    if verbose:
        print(f"✅ {model_type.upper()} model trained on {len(X_train)} rows")
        print(f"📈 Forecasting {len(y_pred)} steps ahead")

    return y_pred, y_test
```

### LSTM Model

```python
import numpy as np
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense, Input
from sklearn.preprocessing import MinMaxScaler

def run_lstm(df_queue, sequence_len=48, verbose=True):
    df_queue = df_queue.copy()

    # Optional: Add holiday flag if available
    if 'is_holiday' in df_queue.columns:
        df_queue['Calls'] = df_queue['Calls'] * (1.2 if df_queue['is_holiday'].any() else 1.0)

    # Step 1: Scale the 'Calls' column
    scaler = MinMaxScaler()
    df_queue['Calls_scaled'] = scaler.fit_transform(df_queue[['Calls']])

    # Step 2: Create sequences
    sequence = df_queue['Calls_scaled'].values
    X, y = [], []
    for i in range(sequence_len, len(sequence)):
        X.append(sequence[i-sequence_len:i])
        y.append(sequence[i])

    if len(X) == 0:
        print(f"⚠️ Skipping LSTM – not enough data for sequence_len={sequence_len}")
        return np.array([]), np.array([])

    X, y = np.array(X), np.array(y)
    X = X.reshape((X.shape[0], X.shape[1], 1))

    # Step 3: Build and train the LSTM model
    model = Sequential([
        Input(shape=(sequence_len, 1)),
        LSTM(50, activation='relu'),
        Dense(1)
    ])
    model.compile(optimizer='adam', loss='mse')
    model.fit(X, y, epochs=10, verbose=0)

    # Step 4: Predict and inverse transform
    y_pred_scaled = model.predict(X)
    y_pred = scaler.inverse_transform(y_pred_scaled.reshape(-1, 1)).flatten()
    y_true = scaler.inverse_transform(y.reshape(-1, 1)).flatten()

    if verbose:
        print(f"✅ LSTM trained on {len(X)} sequences")
        print(f"📈 Sequence length: {sequence_len}, Forecast horizon: {len(y_pred)}")

    return y_pred, y_true

```

### Model Comparison

```python
queue_dfs = {q: df[df[q] == 1].copy() for q in queue_cols}
forecast_steps = 48
initial_results = {}

for queue in queue_dfs:
    df_q = queue_dfs[queue]
    initial_results[queue] = {}

    # Prophet
    try:
        pred, actual = run_prophet(df_q, forecast_steps, verbose=False)
        initial_results[queue]['Prophet'] = evaluate(actual['y'][-forecast_steps:], pred['yhat'][-forecast_steps:], verbose=False)
    except: pass

    # ARIMA
    try:
        pred, actual = run_arima(df_q, forecast_steps, verbose=False)
        initial_results[queue]['ARIMA'] = evaluate(actual, pred, verbose=False)
    except: pass

    # XGBoost
    try:
        pred, actual = run_ml_model(df_q, model_type='xgb', verbose=False)
        initial_results[queue]['XGBoost'] = evaluate(actual, pred, verbose=False)
    except: pass

    # Random Forest
    try:
        pred, actual = run_ml_model(df_q, model_type='rf', verbose=False)
        initial_results[queue]['RandomForest'] = evaluate(actual, pred, verbose=False)
    except: pass

    # LSTM
    try:
        pred, actual = run_lstm(df_q, sequence_len=forecast_steps, verbose=False)
        initial_results[queue]['LSTM'] = evaluate(actual, pred, verbose=False)
    except: pass
```

```python
def generate_model_summary_by_mape(results, title="Enhanced Model Performance Summary by Queue (Based on MAPE)"):
    from IPython.display import Markdown, display

    header = f"## 🏆 {title}\n\n"
    table = "| Queue       | Model          | MAE   | RMSE  | MAPE (%) | 🥇 Best Model |\n"
    table += "|-------------|----------------|-------|-------|-----------|---------------|\n"

    for queue, metrics in results.items():
        best_model = min(metrics.items(), key=lambda x: x[1][2])[0]
        for i, (model_name, (mae, rmse, mape)) in enumerate(metrics.items()):
            queue_label = f"**{queue}**" if i == 0 else ""
            best_flag = "✅" if model_name == best_model else "❌"
            table += f"| {queue_label:<11} | {model_name:<14} | {mae:.2f} | {rmse:.2f} | {mape:.2f}     | {best_flag} |\n"

    display(Markdown(header + table))
```

## 📈 Initial Forecast vs Actual Plots

```python
# Select best models ranked by MAPE
def get_best_models_by_mape(results):
    return {
        queue: sorted(metrics.items(), key=lambda x: x[1][2])
        for queue, metrics in results.items()
    }

# Plot forecast vs actual using fallback and last 50 intervals
def plot_best_forecast_with_fallback(queue_name, df_queue, model_ranked, forecast_steps=48):
    def has_enough_data(df, threshold=10):
        return len(df.dropna()) >= threshold

    for model_name, _ in model_ranked:
        print(f"\n📈 Trying {model_name} for {queue_name}…")
        try:
            if model_name == 'Prophet':
                pred, actual = run_prophet(df_queue, forecast_steps, verbose=False)
                y_true = actual['y'][-forecast_steps:]
                y_pred = pred['yhat'][-forecast_steps:]
                time_index = actual.index[-forecast_steps:]

            elif model_name == 'ARIMA':
                y_pred, y_true = run_arima(df_queue, forecast_steps, verbose=False)
                time_index = df_queue['Date_Time'].dropna().reset_index(drop=True).iloc[-forecast_steps:]

            elif model_name == 'XGBoost':
                y_pred, y_true = run_ml_model(df_queue, model_type='xgb', verbose=False)
                if not has_enough_data(df_queue): raise ValueError("Insufficient data")
                time_index = df_queue['Date_Time'].dropna().reset_index(drop=True).iloc[-len(y_pred):]

            elif model_name == 'RandomForest':
                y_pred, y_true = run_ml_model(df_queue, model_type='rf', verbose=False)
                if not has_enough_data(df_queue): raise ValueError("Insufficient data")
                time_index = df_queue['Date_Time'].dropna().reset_index(drop=True).iloc[-len(y_pred):]

            elif model_name == 'LSTM':
                y_pred, y_true = run_lstm(df_queue, sequence_len=forecast_steps, verbose=False)
                if not has_enough_data(df_queue): raise ValueError("Insufficient data")
                time_index = df_queue['Date_Time'].dropna().reset_index(drop=True).iloc[-len(y_pred):]

            else:
                print(f"⚠️ Unknown model: {model_name}")
                continue

            # Trim to last 50 intervals
            plot_len = min(50, len(y_pred), len(y_true), len(time_index))
            y_true = y_true[-plot_len:]
            y_pred = y_pred[-plot_len:]
            time_index = time_index[-plot_len:]

            # Plot
            plt.figure(figsize=(12, 5))
            plt.plot(time_index, y_true, label='Actual', linewidth=2)
            plt.plot(time_index, y_pred, label='Forecast', linestyle='--')
            plt.title(f"{queue_name} – {model_name} Forecast vs Actual (Last {plot_len} Intervals)")
            plt.xlabel('Date Time')
            plt.ylabel('Call Volume')
            plt.legend()
            plt.xticks(rotation=45)
            plt.tight_layout()
            plt.show()
            return  # Success

        except Exception as e:
            print(f"❌ {model_name} failed or had insufficient data: {e}")
            continue

    print(f"⚠️ No valid model could be plotted for {queue_name}")

# Run for all queues
ranked_models = get_best_models_by_mape(initial_results)

for queue in ranked_models:
    df_q = queue_dfs[queue]
    plot_best_forecast_with_fallback(queue, df_q, ranked_models[queue])
```

### Adding New Features

```python
def add_all_features(df):
    df = df.copy()

    # Ensure datetime format and sort
    df['Date_Time'] = pd.to_datetime(df['Date_Time'])
    df = df.sort_values(by='Date_Time')

    # === Lag features ===
    for lag in [1, 2, 7, 14]:
        df[f'lag_{lag}'] = df['Calls'].shift(lag)

    # === Rolling statistics ===
    for window in [7, 14, 30]:
        df[f'rolling_mean_{window}'] = df['Calls'].rolling(window=window).mean()
        df[f'rolling_std_{window}'] = df['Calls'].rolling(window=window).std()

    # === Holiday impact ===
    df['is_holiday'] = df['is_holiday'].fillna(0).astype(int)
    df['lagged_holiday_1'] = df['is_holiday'].shift(1).fillna(0).astype(int)
    df['lagged_holiday_2'] = df['is_holiday'].shift(2).fillna(0).astype(int)

    # === Time-of-day buckets ===
    def time_bucket(hour):
        if 0 <= hour < 6:
            return 'overnight'
        elif 6 <= hour < 12:
            return 'morning'
        elif 12 <= hour < 18:
            return 'afternoon'
        else:
            return 'evening'

    df['time_of_day'] = df['hour'].apply(time_bucket)
    df = pd.get_dummies(df, columns=['time_of_day'], drop_first=False)

    return df
```

```python
def compare_full_metrics(initial, enhanced):
    from IPython.display import Markdown, display

    header = "## 🔍 Full Comparison of Initial vs Enhanced Model Results\n\n"
    table = "| Queue       | Model          | MAE Init | MAE Enh | Δ MAE | RMSE Init | RMSE Enh | Δ RMSE | MAPE Init | MAPE Enh | Δ MAPE |\n"
    table += "|-------------|----------------|----------|---------|--------|-----------|----------|--------|------------|-----------|---------|\n"

    for queue in initial:
        for model in initial[queue]:
            mae_i, rmse_i, mape_i = initial[queue][model]
            mae_e, rmse_e, mape_e = enhanced.get(queue, {}).get(model, (None, None, None))

            if None in (mae_e, rmse_e, mape_e):
                delta_mae = delta_rmse = delta_mape = "N/A"
            else:
                delta_mae = f"{mae_e - mae_i:+.2f}"
                delta_rmse = f"{rmse_e - rmse_i:+.2f}"
                delta_mape = f"{mape_e - mape_i:+.2f}"

            queue_label = f"**{queue}**" if model == list(initial[queue].keys())[0] else ""
            table += f"| {queue_label:<11} | {model:<14} | {mae_i:.2f}    | {mae_e:.2f}   | {delta_mae} | {rmse_i:.2f}     | {rmse_e:.2f}    | {delta_rmse} | {mape_i:.2f}      | {mape_e:.2f}     | {delta_mape} |\n"

    display(Markdown(header + table))
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

def plot_mape_heatmap(initial, enhanced):
    queues = list(initial.keys())
    models = ['Prophet', 'ARIMA', 'XGBoost', 'RandomForest', 'LSTM']
    delta_matrix = pd.DataFrame(index=queues, columns=models)

    for queue in queues:
        for model in models:
            if model in initial[queue] and model in enhanced.get(queue, {}):
                mape_i = initial[queue][model][2]
                mape_e = enhanced[queue][model][2]
                delta = mape_e - mape_i
                delta_matrix.loc[queue, model] = delta
            else:
                delta_matrix.loc[queue, model] = np.nan

    plt.figure(figsize=(12, len(queues) * 0.6 + 2))
    sns.heatmap(delta_matrix.astype(float), annot=True, fmt=".2f", cmap="RdYlGn", center=0, linewidths=0.5, cbar_kws={'label': 'Δ MAPE'})
    plt.title("🔍 Heatmap of MAPE Improvements (Enhanced – Initial)")
    plt.xlabel("Model")
    plt.ylabel("Queue")
    plt.tight_layout()
    plt.show()
```

## Make the Machine Self Aware - Selecting best Features

### Modular Feature Builder

```python
def add_feature_group(df, group):
    df = df.copy()
    df['Date_Time'] = pd.to_datetime(df['Date_Time'])
    df = df.sort_values(by='Date_Time')

    if group == 'lags':
        for lag in [1, 2, 7, 14]:
            df[f'lag_{lag}'] = df['Calls'].shift(lag)

    elif group == 'rolling':
        for window in [7, 14, 30]:
            df[f'rolling_mean_{window}'] = df['Calls'].rolling(window=window).mean()
            df[f'rolling_std_{window}'] = df['Calls'].rolling(window=window).std()

    elif group == 'holiday':
        df['is_holiday'] = df['is_holiday'].fillna(0).astype(int)
        df['lagged_holiday_1'] = df['is_holiday'].shift(1).fillna(0).astype(int)
        df['lagged_holiday_2'] = df['is_holiday'].shift(2).fillna(0).astype(int)

    elif group == 'time_bucket':
        def time_bucket(hour):
            if 0 <= hour < 6: return 'overnight'
            elif 6 <= hour < 12: return 'morning'
            elif 12 <= hour < 18: return 'afternoon'
            else: return 'evening'
        df['time_of_day'] = df['hour'].apply(time_bucket)
        df = pd.get_dummies(df, columns=['time_of_day'], drop_first=False)

    return df
```

```python
def evaluate_feature_groups_all_models(df_queue, forecast_steps=48):
    model_types = ['prophet', 'arima', 'xgb', 'rf', 'lstm']
    feature_groups = ['lags', 'rolling', 'holiday', 'time_bucket']
    results = {}

    for model_type in model_types:
        model_results = {}
        try:
            if model_type == 'prophet':
                pred, actual = run_prophet(df_queue, forecast_steps, verbose=False)
                base_mape = evaluate(actual['y'][-forecast_steps:], pred['yhat'][-forecast_steps:], verbose=False)[2]
            elif model_type == 'arima':
                pred, actual = run_arima(df_queue, forecast_steps, verbose=False)
                base_mape = evaluate(actual, pred, verbose=False)[2]
            elif model_type == 'xgb':
                pred, actual = run_ml_model(df_queue, model_type='xgb', verbose=False)
                base_mape = evaluate(actual, pred, verbose=False)[2]
            elif model_type == 'rf':
                pred, actual = run_ml_model(df_queue, model_type='rf', verbose=False)
                base_mape = evaluate(actual, pred, verbose=False)[2]
            elif model_type == 'lstm':
                pred, actual = run_lstm(df_queue, sequence_len=forecast_steps, verbose=False)
                base_mape = evaluate(actual, pred, verbose=False)[2]
            else:
                continue
        except:
            base_mape = None

        model_results['base'] = base_mape

        for group in feature_groups:
            try:
                df_enhanced = add_feature_group(df_queue.copy(), group)
                if model_type == 'prophet':
                    pred, actual = run_prophet(df_enhanced, forecast_steps, verbose=False)
                    mape = evaluate(actual['y'][-forecast_steps:], pred['yhat'][-forecast_steps:], verbose=False)[2]
                elif model_type == 'arima':
                    pred, actual = run_arima(df_enhanced, forecast_steps, verbose=False)
                    mape = evaluate(actual, pred, verbose=False)[2]
                elif model_type == 'xgb':
                    pred, actual = run_ml_model(df_enhanced, model_type='xgb', verbose=False)
                    mape = evaluate(actual, pred, verbose=False)[2]
                elif model_type == 'rf':
                    pred, actual = run_ml_model(df_enhanced, model_type='rf', verbose=False)
                    mape = evaluate(actual, pred, verbose=False)[2]
                elif model_type == 'lstm':
                    pred, actual = run_lstm(df_enhanced, sequence_len=forecast_steps, verbose=False)
                    mape = evaluate(actual, pred, verbose=False)[2]
                else:
                    mape = None
            except:
                mape = None

            delta = mape - base_mape if base_mape is not None and mape is not None else "N/A"
            model_results[group] = {'MAPE': mape, 'Δ MAPE': delta}

        results[model_type] = model_results

    return results
```

```python
full_feature_impact = {}

for queue in queue_cols:
    df_q = df[df[queue] == 1].copy()
    print(f"\n🔍 Evaluating {queue} across all models…")
    full_feature_impact[queue] = evaluate_feature_groups_all_models(df_q)
```

### Self-Aware Feature Selection + Reevaluation

### Feature Selection Map

```python
selected_features_map = {}

for queue in queue_cols:
    df_q = df[df[queue] == 1].copy()
    results = evaluate_feature_groups_all_models(df_q)

    selected_features_map[queue] = {}

    for model, metrics in results.items():
        base_mape = metrics['base']
        selected = []
        for group, result in metrics.items():
            if group == 'base': continue
            if isinstance(result['Δ MAPE'], float) and result['Δ MAPE'] < 0:
                selected.append(group)
        selected_features_map[queue][model] = selected
```

### Smart Feature Builder

```python
def add_selected_features(df, selected_groups):
    df = df.copy()
    df['Date_Time'] = pd.to_datetime(df['Date_Time'])
    df = df.sort_values(by='Date_Time')

    for group in selected_groups:
        df = add_feature_group(df, group)

    return df
```

### Rerun Benchmarking with Selected Features

```python
smart_results = {}

for queue in queue_cols:
    df_q_raw = df[df[queue] == 1].copy()
    smart_results[queue] = {}

    for model in selected_features_map[queue]:
        selected_groups = selected_features_map[queue][model]
        df_q = add_selected_features(df_q_raw, selected_groups)

        try:
            if model == 'prophet':
                pred, actual = run_prophet(df_q, forecast_steps=48, verbose=False)
                mae, rmse, mape = evaluate(actual['y'][-48:], pred['yhat'][-48:], verbose=False)

            elif model == 'arima':
                pred, actual = run_arima(df_q, forecast_steps=48, verbose=False)
                mae, rmse, mape = evaluate(actual, pred, verbose=False)

            elif model == 'xgb':
                pred, actual = run_ml_model(df_q, model_type='xgb', verbose=False)
                mae, rmse, mape = evaluate(actual, pred, verbose=False)

            elif model == 'rf':
                pred, actual = run_ml_model(df_q, model_type='rf', verbose=False)
                mae, rmse, mape = evaluate(actual, pred, verbose=False)

            elif model == 'lstm':
                pred, actual = run_lstm(df_q, sequence_len=48, verbose=False)
                mae, rmse, mape = evaluate(actual, pred, verbose=False)

            else:
                continue

            smart_results[queue][model] = (mae, rmse, mape)

        except Exception as e:
            smart_results[queue][model] = None
```

```python
def generate_full_best_model_summary(smart_results):
    from IPython.display import Markdown, display

    header = "## 🏆 Smart Model Performance Summary by Queue (MAE, RMSE, MAPE)\n\n"
    table = "| Queue        | Best Model     | MAE   | RMSE  | MAPE (%) |\n"
    table += "|--------------|----------------|-------|-------|-----------|\n"

    for queue, model_scores in smart_results.items():
        best_model = None
        best_metrics = (None, None, float('inf'))  # MAE, RMSE, MAPE

        for model, metrics in model_scores.items():
            if metrics is None or not isinstance(metrics, (tuple, list)) or len(metrics) != 3:
                continue
            mae, rmse, mape = metrics
            if mape < best_metrics[2]:
                best_model = model
                best_metrics = (mae, rmse, mape)

        if best_model:
            table += f"| {queue:<13} | {best_model:<14} | {best_metrics[0]:.2f} | {best_metrics[1]:.2f} | {best_metrics[2]:.2f}     |\n"

    display(Markdown(header + table))
```

```python
def plot_forecast_dynamic(queue_name, df_queue, smart_results, forecast_steps=48):
    def has_enough_data(df, threshold=10):
        return len(df.dropna()) >= threshold

    # Rank models by MAPE
    model_metrics = smart_results.get(queue_name, {})
    ranked_models = sorted(
        [(model, metrics[2]) for model, metrics in model_metrics.items() if metrics is not None],
        key=lambda x: x[1]
    )

    for model_name, _ in ranked_models:
        print(f"\n📈 Trying {model_name} for {queue_name}…")
        try:
            if model_name == 'prophet':
                pred, actual = run_prophet(df_queue, forecast_steps, verbose=False)
                y_true = actual['y'][-forecast_steps:]
                y_pred = pred['yhat'][-forecast_steps:]
                time_index = actual.index[-forecast_steps:]

            elif model_name == 'arima':
                y_pred, y_true = run_arima(df_queue, forecast_steps, verbose=False)
                time_index = df_queue['Date_Time'].dropna().reset_index(drop=True).iloc[-forecast_steps:]

            elif model_name == 'xgb':
                y_pred, y_true = run_ml_model(df_queue, model_type='xgb', verbose=False)
                if not has_enough_data(df_queue): raise ValueError("Insufficient data")
                time_index = df_queue['Date_Time'].dropna().reset_index(drop=True).iloc[-len(y_pred):]

            elif model_name == 'rf':
                y_pred, y_true = run_ml_model(df_queue, model_type='rf', verbose=False)
                if not has_enough_data(df_queue): raise ValueError("Insufficient data")
                time_index = df_queue['Date_Time'].dropna().reset_index(drop=True).iloc[-len(y_pred):]

            elif model_name == 'lstm':
                y_pred, y_true = run_lstm(df_queue, sequence_len=forecast_steps, verbose=False)
                if not has_enough_data(df_queue): raise ValueError("Insufficient data")
                time_index = df_queue['Date_Time'].dropna().reset_index(drop=True).iloc[-len(y_pred):]

            else:
                print(f"⚠️ Unknown model: {model_name}")
                continue

            # Trim to last 50 intervals
            plot_len = min(50, len(y_pred), len(y_true), len(time_index))
            y_true = y_true[-plot_len:]
            y_pred = y_pred[-plot_len:]
            time_index = time_index[-plot_len:]

            # Plot
            plt.figure(figsize=(12, 5))
            plt.plot(time_index, y_true, label='Actual', linewidth=2)
            plt.plot(time_index, y_pred, label='Forecast', linestyle='--')
            plt.title(f"{queue_name} – {model_name} Forecast vs Actual (Last {plot_len} Intervals)")
            plt.xlabel('Date Time')
            plt.ylabel('Call Volume')
            plt.legend()
            plt.xticks(rotation=45)
            plt.tight_layout()
            plt.show()
            return  # Success

        except Exception as e:
            print(f"❌ {model_name} failed or had insufficient data: {e}")
            continue

    print(f"⚠️ No valid model could be plotted for {queue_name}")
```

```python
for queue in queue_cols:
    df_q = df[df[queue] == 1].copy()
    plot_forecast_dynamic(queue, df_q, smart_results)
```
